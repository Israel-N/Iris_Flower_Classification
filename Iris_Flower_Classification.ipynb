# Iris Flower Classification

**Objective:** Classify Iris flowers into **Setosa, Versicolor, and Virginica** using sepal and petal measurements.

### Project workflow
1. Load the Iris dataset
2. Explore and visualize the data
3. Check data quality
4. Split into training and testing sets
5. Standardize the features
6. Train a Logistic Regression classifier
7. Evaluate the model
8. Visualize the confusion matrix
9. Predict a new flower
10. Save the trained model


```python
#Import libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully.")
```

```python
#Load the Iris dataset

iris = load_iris()

X = iris.data
y = iris.target

print("Dataset loaded successfully!")
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Feature names:", iris.feature_names)
print("Class names:", iris.target_names)
```

```python
#Create a Pandas DataFrame

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target

df["species"] = df["target"].map({
    0: "Setosa",
    1: "Versicolor",
    2: "Virginica"
})

df.head()
```

```python
#Explore the dataset

print("Dataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python
#Check for missing values

print("Missing values in each column:")
print(df.isnull().sum())
```

```python
#Check class distribution

print(df["species"].value_counts())

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="species")
plt.title("Number of Samples per Iris Species")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.show()
```

```python
#Scatter plot using petal measurements

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species",
    s=100
)

plt.title("Iris Species Based on Petal Measurements")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python
#Scatter plot using sepal measurements

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="sepal width (cm)",
    hue="species",
    s=100
)

plt.title("Iris Species Based on Sepal Measurements")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

```python
#Feature histograms

df[iris.feature_names].hist(
    figsize=(12, 8),
    bins=15
)

plt.suptitle("Distribution of Iris Features")
plt.tight_layout()
plt.show()
```

```python
#Pair plot

sns.pairplot(
    df,
    vars=iris.feature_names,
    hue="species",
    diag_kind="hist"
)

plt.show()
```

## Preparing the data

The four measurements will be used as input features:

- Sepal length
- Sepal width
- Petal length
- Petal width

The target is the Iris species.


```python
#Define features and target

feature_columns = iris.feature_names

X = df[feature_columns]
y = df["target"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())
```

```python
#Split the data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
#Standardize the features

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## Model training

We will use **Logistic Regression** as the classification algorithm.

The model will learn the relationship between the four flower measurements and the three Iris species.


```python
#Create and train the Logistic Regression model

model = LogisticRegression(
    max_iter=200
)

model.fit(
    X_train_scaled,
    y_train
)

print("Model training completed successfully!")
```

```python
#Make predictions

y_pred = model.predict(X_test_scaled)

print("Predicted classes:")
print(y_pred)

print("\nActual classes:")
print(y_test.to_numpy())
```

```python
#Calculate evaluation metrics

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted"
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted"
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted"
)

print("Model Evaluation")
print("=" * 40)
print(f"Accuracy :  {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision:  {precision:.4f}")
print(f"Recall   :  {recall:.4f}")
print(f"F1 Score :  {f1:.4f}")
```

```python
#Classification report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)
```

```python
#Confusion matrix

cm = confusion_matrix(
    y_test,
    y_pred
)

print("Confusion Matrix:")
print(cm)
```

```python
# Visualize the confusion matrix

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title("Iris Flower Classification - Confusion Matrix")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.tight_layout()
plt.show()
```

## Predict a new Iris flower

The following example uses:

- Sepal length: 5.1 cm
- Sepal width: 3.5 cm
- Petal length: 1.4 cm
- Petal width: 0.2 cm


```python
# Predict a new flower

new_flower = np.array([
    [5.1, 3.5, 1.4, 0.2]
])

new_flower_scaled = scaler.transform(new_flower)

prediction = model.predict(new_flower_scaled)

predicted_species = iris.target_names[prediction[0]]

print("Predicted species:", predicted_species)
```

```python
#Show prediction probabilities

probabilities = model.predict_proba(
    new_flower_scaled
)

for species, probability in zip(
    iris.target_names,
    probabilities[0]
):
    print(
        f"{species}: {probability * 100:.2f}%"
    )
```

```python
#Save the trained model and scaler
import os

os.makedirs("models", exist_ok=True)

joblib.dump(
    model,
    "models/iris_logistic_regression.pkl"
)

joblib.dump(
    scaler,
    "models/iris_scaler.pkl"
)

print("Model saved to:")
print("models/iris_logistic_regression.pkl")

print("\nScaler saved to:")
print("models/iris_scaler.pkl")
```

```python
#Test loading the saved model

loaded_model = joblib.load(
    "models/iris_logistic_regression.pkl"
)

loaded_scaler = joblib.load(
    "models/iris_scaler.pkl"
)

test_flower = np.array([
    [6.0, 3.0, 4.8, 1.8]
])

test_flower_scaled = loaded_scaler.transform(
    test_flower
)

test_prediction = loaded_model.predict(
    test_flower_scaled
)

print(
    "Loaded model prediction:",
    iris.target_names[test_prediction[0]]
)
```
